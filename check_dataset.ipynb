# Inspect the Apertus dataset
Run from the repository root. Change `SPLIT`, `ROW`, and `QID` to browse examples. Only the tokenizer is loaded; no model or GPU is needed.


```python
import json
from collections import Counter
import os

from transformers import AutoTokenizer

from apertus_data import load_examples
from train_apertus import BASE_MODEL, BASE_REVISION, DATA_DIR

```

```python
tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL, revision=BASE_REVISION)
train = load_examples(os.path.join(DATA_DIR, "train.jsonl"), tokenizer)
```

## Inspect individual tokens
Adjust the slice below. The head scores each option using its end-token hidden state and the decide-token hidden state; cross-entropy compares those scores with the target option index.


```python
import textwrap

token_ids = train[4163]["ids"]
text = tokenizer.decode(
    token_ids,
    skip_special_tokens=False,
)

print("\n".join(textwrap.fill(line, width=100) for line in text.splitlines()))
```
